Learn to create a circuit and apply X and H gates.

```python
from qiskit import QuantumCircuit

qc = QuantumCircuit(1)
qc.x(0)  # |0> -> |1>
qc.h(0)  # create a superposition with a relative phase
print(qc.draw())
```

## Explore
Apply H a second time. What state do you obtain? Why does the phase sign matter?
