Prepare and sample a Bell state using Qiskit Aer.

```python
from qiskit_learning_lab import bell_circuit, simulate_counts

qc = bell_circuit()
print(qc.draw())
print(simulate_counts(qc, shots=4096, seed=21))
```

Ideal expectation: outcomes 00 and 11 occur with probabilities 0.5 each. Finite-shot variation is normal. This is simulation, not hardware evidence.
