Compare sampled distributions with theoretical distributions.

```python
from qiskit_learning_lab import (
    bell_circuit,
    counts_to_probabilities,
    simulate_counts,
    total_variation_distance,
)

counts = simulate_counts(bell_circuit(), shots=4096, seed=21)
observed = counts_to_probabilities(counts)
ideal = {"00": 0.5, "11": 0.5}
print("Observed:", observed)
print("TVD:", total_variation_distance(observed, ideal))
```

TVD measures the difference between two discrete probability distributions. Add a documented Aer noise model as an extension and label it as simulated noise, not physical device calibration.
